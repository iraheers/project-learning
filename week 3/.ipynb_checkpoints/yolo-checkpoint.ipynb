# With Food Dataset

```python
from ultralytics import YOLO

# Load a pretrained YOLO11n model
model = YOLO("yolo11n.pt")

# Train the model on the COCO8 dataset for 100 epochs
train_results = model.train(
    data="./food_dataset/yolo.yaml",  # Path to dataset configuration file
    epochs=10,  # Number of training epochs
    imgsz=640,  # Image size for training
    device="mps",  # Device to run on (e.g., 'cpu', 0, [0,1,2,3])
)

# Evaluate the model's performance on the validation set
metrics = model.val()
```

```python
# Perform object detection on an image
results = model("./test_images/plate1.jpeg")  # Predict on an image
results[0].show()  # Display results
```

```python
# Perform object detection on an image
results = model("./test_images/fruits.jpg")  # Predict on an image
results[0].show()  # Display results
```
